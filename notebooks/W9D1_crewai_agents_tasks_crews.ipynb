
# W9D1 — CrewAI Fundamentals: Agents, Tasks & Crews

## Objective
Build and evaluate a local multi-agent research workflow using CrewAI and Ollama.

## Agents
1. Researcher — gathers relevant information.
2. Writer — creates a structured research report.
3. Reviewer — evaluates evidence, accuracy, completeness, and clarity.

## Workflow
Research → Writing → Review → Quality evaluation → Web-search-enhanced research

## Model
- Provider: Ollama
- Model: qwen2.5:3b

## Evaluation criteria
- Clear research question and scope
- Evidence and source attribution
- Separation of facts from limitations or uncertainty
- Complete, well-structured report
- Reviewer feedback and documented quality comparison

Record actual outputs, validation results, and any limitations.
