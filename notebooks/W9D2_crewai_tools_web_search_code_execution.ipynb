```python
from crewai.tools import BaseTool
from ddgs import DDGS
import json

class WebSearchTool(BaseTool):
    name: str = "DuckDuckGo Web Search"
    description: str = (
        "Search the live web for a research query and return result titles, URLs, and snippets."
    )

    def _run(self, query: str) -> str:
        try:
            results = list(DDGS().text(query, max_results=5))
            if not results:
                return "No search results found."
            return json.dumps(
                [
                    {
                        "title": item.get("title", ""),
                        "url": item.get("href", ""),
                        "snippet": item.get("body", ""),
                    }
                    for item in results
                ],
                ensure_ascii=False,
                indent=2,
            )
        except Exception as exc:
            return f"Search failed: {type(exc).__name__}: {exc}"

web_search = WebSearchTool()

print(web_search.run(query="Python official ast documentation"))
```

```python
import json
from crewai.tools import BaseTool
from ddgs import DDGS

class WebSearchTool(BaseTool):
    name: str = "DuckDuckGo Web Search"
    description: str = "Search the live web and return useful titles, URLs, and snippets."

    def _run(self, query: str) -> str:
        try:
            with DDGS() as ddgs:
                results = list(
                    ddgs.text(query, backend="duckduckgo", max_results=5)
                )

            usable = [
                {
                    "title": r["title"],
                    "url": r["href"],
                    "snippet": r["body"],
                }
                for r in results
                if r.get("title") and r.get("href")
            ]

            return json.dumps(usable, ensure_ascii=False, indent=2)

        except Exception as exc:
            return f"Search failed: {type(exc).__name__}: {exc}"

web_search = WebSearchTool()
print(web_search.run(query="Python official ast documentation"))
```

```python
tool_researcher = Agent(
    role="Web Research Analyst",
    goal="Find reliable, current information and support findings with real sources.",
    backstory="You are a careful researcher who checks search results, records source URLs, and distinguishes evidence from assumptions.",
    tools=[web_search],
    llm=llm,
    verbose=True,
)

print("Researcher created with live web-search access!")
```

```python
from crewai import Task

web_research_task = Task(
    description=(
        "Research the practical uses and limitations of Python Abstract Syntax Trees (AST). "
        "Use the DuckDuckGo Web Search tool to find the official Python AST documentation "
        "and one additional credible source. Include key findings, exact source URLs, "
        "and limitations. Do not invent sources or claim that snippets prove more than they do."
    ),
    expected_output=(
        "Research notes containing findings, at least two real source URLs when available, "
        "and clearly stated limitations."
    ),
    agent=tool_researcher,
)

print("Web-enhanced research task created!")
```

```python
web_writer = Agent(
    role="Evidence-Based Technical Writer",
    goal="Write a clear report supported by the research findings and source URLs.",
    backstory="You write structured reports and never invent citations.",
    llm=llm,
    verbose=True,
)

web_reviewer = Agent(
    role="Research Quality Reviewer",
    goal="Evaluate accuracy, evidence, completeness, and clarity.",
    backstory="You check whether claims are supported by the supplied research and flag missing evidence.",
    llm=llm,
    verbose=True,
)

web_writing_task = Task(
    description=(
        "Write a structured report about Python AST practical uses and limitations "
        "using the research notes. Include an introduction, use cases, limitations, "
        "conclusion, and a Sources section with the exact URLs supplied by the researcher."
    ),
    expected_output="A concise technical report with source URLs and clear limitations.",
    agent=web_writer,
    context=[web_research_task],
)

web_review_task = Task(
    description=(
        "Review the report for accuracy, evidence quality, completeness, and clarity. "
        "Score each criterion out of 5. Check whether the listed URLs were supplied "
        "in the research notes, and identify unsupported claims and improvements."
    ),
    expected_output="A quality review with four scores and actionable feedback.",
    agent=web_reviewer,
    context=[web_research_task, web_writing_task],
)

web_crew = Crew(
    agents=[tool_researcher, web_writer, web_reviewer],
    tasks=[web_research_task, web_writing_task, web_review_task],
    process=Process.sequential,
    verbose=True,
)

print("Web-enhanced crew connected: Research → Writing → Review")
```

```python
import ast
import operator
from crewai.tools import BaseTool

class SafeCalculationTool(BaseTool):
    name: str = "Safe Python Calculator"
    description: str = "Calculate basic arithmetic expressions such as (1200 * 0.15) + 40."

    def _run(self, expression: str) -> str:
        operations = {
            ast.Add: operator.add,
            ast.Sub: operator.sub,
            ast.Mult: operator.mul,
            ast.Div: operator.truediv,
        }

        try:
            tree = ast.parse(expression, mode="eval")

            def calculate(node):
                if isinstance(node, ast.Expression):
                    return calculate(node.body)
                if isinstance(node, ast.Constant) and type(node.value) in (int, float):
                    return node.value
                if isinstance(node, ast.BinOp) and type(node.op) in operations:
                    return operations[type(node.op)](
                        calculate(node.left), calculate(node.right)
                    )
                raise ValueError("Unsupported expression")

            return str(calculate(tree))

        except Exception as exc:
            return f"Rejected: {type(exc).__name__}: {exc}"

safe_calculator = SafeCalculationTool()

print("Calculation:", safe_calculator.run(expression="(1200 * 0.15) + 40"))
print("Safety test:", safe_calculator.run(expression="__import__('os').getcwd()"))
```

```python
tool_researcher.tools.append(safe_calculator)

print("Researcher tools:")
for tool in tool_researcher.tools:
    print("-", tool.name)
```
