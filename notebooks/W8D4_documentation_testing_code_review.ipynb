```python
def calculate_average(values):
    """Return the average of a non-empty list of numbers."""
    if not values:
        raise ValueError("values must not be empty")

    if not all(isinstance(value, (int, float)) for value in values):
        raise TypeError("all values must be numbers")

    return sum(values) / len(values)


# Basic example
scores = [80, 90, 100]
print("Scores:", scores)
print("Average:", calculate_average(scores))
```

```python
# Test normal input
assert calculate_average([80, 90, 100]) == 90.0

# Test a single value
assert calculate_average([50]) == 50.0

# Test decimal values
assert calculate_average([1.5, 2.5]) == 2.0

# Test empty input
try:
    calculate_average([])
    raise AssertionError("Expected ValueError for empty input")
except ValueError as error:
    assert str(error) == "values must not be empty"

# Test invalid input
try:
    calculate_average([10, "20"])
    raise AssertionError("Expected TypeError for invalid input")
except TypeError as error:
    assert str(error) == "all values must be numbers"

print("All 5 tests passed.")
```
