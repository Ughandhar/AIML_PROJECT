```python
import tiktoken

enc = tiktoken.encoding_for_model("gpt-4o")

prompts = {
    "Small": "Explain LiDAR in robotics.",
    
    "Medium": "Explain how LiDAR is used for perception, mapping, localization, and obstacle avoidance in autonomous robots.",
    
    "Large": "Explain LiDAR in robotics, including how laser pulses measure distance, how point clouds are generated, and how the data supports perception, mapping, localization, obstacle detection, SLAM, and autonomous navigation.",
    
    "RAG": "You are a robotics assistant. Explain how LiDAR, cameras, radar, and ultrasonic sensors contribute to robot perception. Compare their strengths, limitations, typical use cases, and how sensor fusion combines their information for reliable navigation and obstacle avoidance.",
    
    "Very Large": "You are an expert robotics assistant. Provide a detailed explanation of autonomous robot perception and navigation. Discuss LiDAR, cameras, radar, ultrasonic sensors, sensor fusion, SLAM, localization, mapping, obstacle detection, path planning, motion planning, and control. Explain how these components work together in an autonomous mobile robot and include practical examples."
}

print("Prompt Token Audit")
print("-" * 50)

for name, prompt in prompts.items():
    token_count = len(enc.encode(prompt))
    print(f"{name:12} : {token_count:4} tokens")
```

```python
input_price_per_1m = 5.00

print("GPT-4o Input Cost Audit")
print("-" * 50)

for name, prompt in prompts.items():
    token_count = len(enc.encode(prompt))
    cost = (token_count / 1_000_000) * input_price_per_1m
    
    print(
        f"{name:12} : "
        f"{token_count:4} tokens | "
        f"${cost:.8f}"
    )
```

```python
from llmlingua import PromptCompressor

print("LLMLingua imported successfully")
```

```python
compressor = PromptCompressor(
    model_name="microsoft/llmlingua-2-bert-base-multilingual-cased-meetingbank",
    use_llmlingua2=True,
    device_map="cpu"
)

print("LLMLingua-2 initialized successfully")
```

```python
rag_prompt = """
You are an expert robotics assistant helping with autonomous mobile robot perception and navigation.

The robot uses multiple sensors to understand its environment. LiDAR measures distance by emitting laser pulses and measuring the return time. It produces a point cloud containing three-dimensional information about surrounding objects. LiDAR is useful for obstacle detection, mapping, localization, and SLAM. It performs well in many lighting conditions, although reflective, transparent, or very dark surfaces can create challenges.

Cameras provide rich visual information such as color, texture, object appearance, lane markings, signs, and semantic information. Cameras are relatively inexpensive compared with many LiDAR systems, but their performance can be affected by lighting, shadows, glare, rain, fog, and poor visibility. Computer vision algorithms can detect and classify objects from camera images.

Radar measures objects using radio waves. It can provide range and relative velocity information and can work in conditions where cameras may struggle, such as darkness, rain, or fog. However, radar generally provides less detailed spatial information than LiDAR and cameras.

Ultrasonic sensors are commonly used for short-range obstacle detection. They are inexpensive and useful for detecting nearby objects during low-speed navigation and parking-like maneuvers. Their effective range is limited and their measurements can be affected by the shape and material of an object.

Sensor fusion combines information from multiple sensors to produce a more reliable representation of the environment. For example, LiDAR can provide accurate geometric information while cameras provide semantic information. Radar can provide velocity information and improve robustness in difficult weather. Combining these measurements can reduce the weaknesses of individual sensors.

SLAM stands for simultaneous localization and mapping. A mobile robot uses sensor observations to estimate its position while constructing or updating a map of the environment. LiDAR-based SLAM commonly extracts geometric information from consecutive scans and estimates the robot's motion. Visual SLAM uses camera images and visual features. Sensor fusion can combine visual, inertial, LiDAR, and other measurements.

Localization determines the robot's position within a known map. GPS can provide global positioning outdoors, while inertial measurement units provide acceleration and angular velocity. Wheel odometry estimates movement from wheel rotations but can accumulate error due to wheel slip. A Kalman filter or related state-estimation method can combine noisy measurements from different sensors.

Path planning determines a suitable route from the robot's current position to a goal. Algorithms such as A* can search a known graph or grid for a path. Sampling-based planners such as RRT can be useful in higher-dimensional configuration spaces. Local planning considers nearby obstacles and robot motion constraints.

Motion planning must consider the robot's physical limitations, including maximum velocity, acceleration, turning radius, and collision constraints. The planner should generate a feasible trajectory rather than only a geometric path.

Control algorithms are responsible for following the planned trajectory. PID control can use the error between the desired and measured state to calculate corrective commands. More advanced methods such as model predictive control can optimize future control actions while considering system constraints.

Obstacle detection identifies objects that may interfere with the robot's motion. LiDAR point clouds, camera detections, radar targets, and ultrasonic measurements can all contribute. A perception system may track detected objects over time and estimate their position and velocity.

A complete autonomous navigation pipeline commonly contains sensing, preprocessing, perception, localization, mapping, planning, and control. Sensor data is first collected and filtered. Perception algorithms detect relevant environmental features and objects. Localization estimates the robot state. Mapping maintains an environment representation. Planning selects a route and trajectory. Finally, the controller sends commands to the motors.

Consider a warehouse robot navigating between storage shelves. LiDAR can detect shelf boundaries and obstacles, cameras can identify labels and objects, wheel encoders can estimate motion, and an IMU can provide orientation information. Sensor fusion can combine these measurements. SLAM can maintain the warehouse map, localization can estimate the robot position, and a path planner can calculate a collision-free route.

Reliable autonomous navigation requires careful handling of uncertainty. Sensor measurements contain noise and may occasionally be incorrect or missing. A robust system should detect invalid measurements, maintain confidence estimates, and avoid depending entirely on one sensor.

In practical robotics systems, the choice of sensors depends on cost, environment, required accuracy, operating speed, safety requirements, and computational resources. A research robot may use LiDAR, multiple cameras, radar, IMU, GPS, and wheel encoders, while a small indoor robot may use cameras, ultrasonic sensors, and wheel odometry.

The final navigation system should integrate perception, localization, planning, and control into a consistent pipeline. The system should continuously update its understanding of the environment and adapt its trajectory when obstacles or unexpected events are detected.
"""

original_tokens = len(enc.encode(rag_prompt))

print(f"Original RAG prompt tokens: {original_tokens}")
```

```python
# Expand the robotics RAG context to approximately 2,000 tokens

rag_prompt_2000 = rag_prompt + "\n\n" + rag_prompt + "\n\n" + rag_prompt[:2500]

original_tokens = len(enc.encode(rag_prompt_2000))

print(f"Expanded RAG prompt tokens: {original_tokens}")
```

```python
compressed_result = compressor.compress_prompt(
    rag_prompt_2000,
    rate=0.4
)

compressed_prompt = compressed_result["compressed_prompt"]

compressed_tokens = len(enc.encode(compressed_prompt))

print("LLMLingua Compression Results")
print("-" * 50)
print(f"Original tokens   : {original_tokens}")
print(f"Compressed tokens : {compressed_tokens}")
print(f"Compression rate  : {compressed_tokens / original_tokens:.2%}")
print(f"Tokens saved      : {original_tokens - compressed_tokens}")
```

```python
compression_summary = {
    "Original tokens": original_tokens,
    "Compressed tokens": compressed_tokens,
    "Tokens saved": original_tokens - compressed_tokens,
    "Reduction percentage": round(
        (1 - compressed_tokens / original_tokens) * 100, 2
    ),
    "Compression rate parameter": 0.4
}

compression_summary
```

```python
key_topics = [
    "LiDAR",
    "camera",
    "radar",
    "ultrasonic",
    "sensor fusion",
    "SLAM",
    "localization",
    "GPS",
    "IMU",
    "Kalman filter",
    "path planning",
    "A*",
    "RRT",
    "PID",
    "obstacle detection",
    "control"
]

original_lower = rag_prompt_2000.lower()
compressed_lower = compressed_prompt.lower()

print("Quality / Information Retention Check")
print("-" * 55)

retained = 0

for topic in key_topics:
    original_has = topic.lower() in original_lower
    compressed_has = topic.lower() in compressed_lower
    
    if original_has and compressed_has:
        status = "Retained"
        retained += 1
    elif original_has and not compressed_has:
        status = "Removed"
    else:
        status = "N/A"
    
    print(f"{topic:20} : {status}")

retention = retained / len(key_topics) * 100

print("-" * 55)
print(f"Key-topic retention: {retention:.2f}%")
```

```python
import chromadb

client = chromadb.Client()

cache_collection = client.get_or_create_collection(
    name="w5d6_semantic_cache",
    metadata={"hnsw:space": "cosine"}
)

print("Semantic cache initialized successfully")
```

```python
qa_pairs = [
    ("What is LiDAR?", "LiDAR is a sensing technology that uses laser pulses to measure distances and create a representation of the environment."),
    ("What is SLAM?", "SLAM stands for Simultaneous Localization and Mapping. It allows a robot to estimate its position while building a map."),
    ("What is sensor fusion?", "Sensor fusion combines measurements from multiple sensors to produce a more reliable estimate of the environment."),
    ("What does a camera provide?", "A camera provides visual information such as color, texture, object appearance, and semantic information."),
    ("What does radar measure?", "Radar can provide range and relative velocity information using radio waves."),
    ("What are ultrasonic sensors used for?", "Ultrasonic sensors are commonly used for short-range obstacle detection."),
    ("What is localization?", "Localization determines the robot's position within a known map."),
    ("What is path planning?", "Path planning determines a suitable route from the robot's current position to a goal."),
    ("What is A*?", "A* is a search algorithm that can find a path through a graph or grid."),
    ("What is RRT?", "RRT is a sampling-based planning algorithm useful for higher-dimensional configuration spaces."),
    ("What is PID control?", "PID control calculates corrective commands using proportional, integral, and derivative error terms."),
    ("What is an IMU?", "An IMU measures quantities such as acceleration and angular velocity."),
    ("What is GPS used for?", "GPS can provide global positioning information, particularly for outdoor robots."),
    ("What is wheel odometry?", "Wheel odometry estimates robot movement using wheel rotation measurements."),
    ("What is a Kalman filter?", "A Kalman filter can combine noisy measurements to estimate the state of a system."),
    ("What is obstacle detection?", "Obstacle detection identifies objects that may interfere with a robot's movement."),
    ("Why combine LiDAR and cameras?", "LiDAR provides geometric information while cameras provide rich visual and semantic information."),
    ("What is motion planning?", "Motion planning generates a feasible trajectory while considering robot physical constraints."),
    ("What is autonomous navigation?", "Autonomous navigation combines sensing, localization, planning, and control to move a robot toward a goal."),
    ("Why is sensor fusion useful?", "Sensor fusion can reduce the weaknesses of individual sensors and improve navigation reliability.")
]

questions = [q for q, a in qa_pairs]
answers = [a for q, a in qa_pairs]

cache_collection.add(
    ids=[f"qa_{i:02d}" for i in range(len(qa_pairs))],
    documents=questions,
    metadatas=[{"answer": answer} for answer in answers]
)

print(f"Q&A pairs stored: {len(qa_pairs)}")
```

```python
query = "How can a robot figure out where it is while creating a map?"

results = cache_collection.query(
    query_texts=[query],
    n_results=1
)

distance = results["distances"][0][0]
similarity = 1 - distance

print("Semantic Cache Query")
print("-" * 50)
print(f"Query      : {query}")
print(f"Similarity : {similarity:.4f}")
print(f"Threshold  : 0.9200")

if similarity >= 0.92:
    print("Cache HIT")
    print(f"Cached answer: {results['metadatas'][0][0]['answer']}")
else:
    print("Cache MISS")
```

```python
from datetime import date

USD_TO_INR = 85.0

requests = [
    {
        "request": "LiDAR explanation",
        "model": "GPT-4o",
        "input_tokens": 500,
        "output_tokens": 150
    },
    {
        "request": "SLAM explanation",
        "model": "GPT-4o",
        "input_tokens": 800,
        "output_tokens": 250
    },
    {
        "request": "Sensor fusion question",
        "model": "GPT-4o mini",
        "input_tokens": 600,
        "output_tokens": 180
    },
    {
        "request": "Path planning question",
        "model": "GPT-4o mini",
        "input_tokens": 450,
        "output_tokens": 120
    },
    {
        "request": "Robotics RAG question",
        "model": "GPT-4o",
        "input_tokens": 1200,
        "output_tokens": 300
    }
]

model_prices = {
    "GPT-4o": {
        "input": 5.00,
        "output": 15.00
    },
    "GPT-4o mini": {
        "input": 0.15,
        "output": 0.60
    }
}

for request in requests:
    prices = model_prices[request["model"]]
    
    usd_cost = (
        request["input_tokens"] / 1_000_000 * prices["input"]
        + request["output_tokens"] / 1_000_000 * prices["output"]
    )
    
    request["usd_cost"] = usd_cost
    request["inr_cost"] = usd_cost * USD_TO_INR
    request["date"] = str(date.today())

print("Daily LLM Cost Tracker")
print("-" * 70)

for request in requests:
    print(
        f"{request['request']:25} | "
        f"{request['model']:12} | "
        f"{request['input_tokens'] + request['output_tokens']:5} tokens | "
        f"₹{request['inr_cost']:.4f}"
    )

daily_total_inr = sum(r["inr_cost"] for r in requests)

print("-" * 70)
print(f"Daily total: ₹{daily_total_inr:.4f}")
```

```python
# Baseline vs optimized RAG cost comparison

baseline_input_tokens = original_tokens
optimized_input_tokens = compressed_tokens

output_tokens = 300

input_price = 5.00
output_price = 15.00

baseline_usd = (
    baseline_input_tokens / 1_000_000 * input_price
    + output_tokens / 1_000_000 * output_price
)

optimized_usd = (
    optimized_input_tokens / 1_000_000 * input_price
    + output_tokens / 1_000_000 * output_price
)

baseline_inr = baseline_usd * USD_TO_INR
optimized_inr = optimized_usd * USD_TO_INR

savings_inr = baseline_inr - optimized_inr
savings_percent = (savings_inr / baseline_inr) * 100

print("Baseline vs Optimized RAG Cost")
print("-" * 55)
print(f"Baseline input tokens : {baseline_input_tokens}")
print(f"Optimized input tokens: {optimized_input_tokens}")
print(f"Output tokens         : {output_tokens}")
print(f"Baseline cost         : ₹{baseline_inr:.4f}")
print(f"Optimized cost        : ₹{optimized_inr:.4f}")
print(f"Cost saved            : ₹{savings_inr:.4f}")
print(f"Cost reduction        : {savings_percent:.2f}%")
```
