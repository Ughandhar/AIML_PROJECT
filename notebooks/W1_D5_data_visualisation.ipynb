```python
# W1D5: Data Visualisation — Matplotlib & Seaborn

## Objective

Use Matplotlib and Seaborn to visualize the dataset and understand distributions, comparisons, correlations, and trends.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
df = pd.read_csv("../data/indian_cities.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)

df.head()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df["city"], df["population_millions"])

plt.title("Population by City")
plt.xlabel("City")
plt.ylabel("Population (Millions)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=df, x="city", y="literacy_rate")

plt.title("Literacy Rate by City")
plt.xlabel("City")
plt.ylabel("Literacy Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=df, x="city", y="average_temperature")

plt.title("Average Temperature by City")
plt.xlabel("City")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="population_millions", bins=6, kde=True)

plt.title("Population Distribution")
plt.xlabel("Population (Millions)")
plt.ylabel("Number of Cities")
plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(8, 5))

numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df.sort_values("population_millions"),
    x="population_millions",
    y="city"
)

plt.title("City Population Comparison")
plt.xlabel("Population (Millions)")
plt.ylabel("City")
plt.tight_layout()
plt.show()
```

```python
## Visualization Findings

1. Delhi has the highest population among the cities in the dataset.
2. Kochi has the lowest population.
3. Chennai has the highest average temperature.
4. Kochi has the highest literacy rate.
5. Jaipur has the lowest literacy rate.
6. The population distribution shows that most cities have populations below 20 million.
7. The correlation heatmap shows weak relationships between the numeric variables.
8. The visualizations make it easier to compare cities and understand patterns in the dataset.
```
