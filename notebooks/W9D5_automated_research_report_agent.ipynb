# W9D5 — Automated Research Report Agent

## Objective
Build an automated research report pipeline by reusing and extending the existing W9D3 research crew.

## Workflow
1. Research the selected topic using web search.
2. Draft a structured research report.
3. Review the report for clarity, evidence, and limitations.
4. Save the report and evaluation evidence.
5. Validate the outputs and document the results.

## Technology Stack
- CrewAI
- Ollama with `qwen2.5:3b`
- DuckDuckGo Search (`ddgs`)
- Python
- JSON and Markdown for evidence and reports
