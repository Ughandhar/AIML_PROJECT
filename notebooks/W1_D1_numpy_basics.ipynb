```python
# 1D Array
array_1d = np.array([10, 20, 30, 40, 50])

# 2D Array
array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 3D Array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:")
print(array_1d)
print("Shape:", array_1d.shape)

print("\n2D Array:")
print(array_2d)
print("Shape:", array_2d.shape)

print("\n3D Array:")
print(array_3d)
print("Shape:", array_3d.shape)
```

```python
# Broadcasting
numbers = np.array([10, 20, 30, 40, 50])
result = numbers + 5

print("Original:", numbers)
print("After adding 5:", result)

# Vectorised operations
values = np.array([1, 2, 3, 4, 5])

squared = values ** 2
doubled = values * 2

print("\nOriginal values:", values)
print("Squared:", squared)
print("Doubled:", doubled)
```

```python
# Matrix multiplication

matrix_a = np.array([
    [1, 2],
    [3, 4]
])

matrix_b = np.array([
    [5, 6],
    [7, 8]
])

matrix_result = matrix_a @ matrix_b

print("Matrix A:")
print(matrix_a)

print("\nMatrix B:")
print(matrix_b)

print("\nMatrix A × B:")
print(matrix_result)
```

```python
from pathlib import Path
import numpy as np

project_root = Path.cwd().parent
csv_path = project_root / "data" / "dataset.csv"

data = np.genfromtxt(
    csv_path,
    delimiter=",",
    skip_header=1
)

print("Dataset:")
print(data)

print("\nShape:", data.shape)
```

```python
# Separate the columns
hours_studied = data[:, 0]
sleep_hours = data[:, 1]
exam_score = data[:, 2]

# Mean
mean_score = np.mean(exam_score)

# Standard deviation
std_score = np.std(exam_score)

# Correlation
correlation = np.corrcoef(hours_studied, exam_score)[0, 1]

print("Mean exam score:", mean_score)
print("Standard deviation:", std_score)
print("Correlation (study hours vs exam score):", correlation)
```
