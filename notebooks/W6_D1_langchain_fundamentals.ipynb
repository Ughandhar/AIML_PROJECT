```python
from langchain_core.prompts import PromptTemplate
from langchain_ollama import OllamaLLM
from langchain_core.output_parsers import StrOutputParser

# Prompt template
prompt = PromptTemplate.from_template(
    "Explain {topic} in simple terms for a beginner."
)

# Local Ollama LLM
llm = OllamaLLM(model="qwen2.5:3b")

# LangChain pipeline:
# PromptTemplate → Ollama LLM → OutputParser
chain = prompt | llm | StrOutputParser()

# Test with 5 inputs
topics = [
    "Artificial Intelligence",
    "Machine Learning",
    "Robotics",
    "Computer Vision",
    "Deep Learning"
]

for topic in topics:
    result = chain.invoke({"topic": topic})
    print(f"\n--- {topic} ---")
    print(result)
```

```python
from langchain_classic.memory import ConversationBufferMemory

# Create conversation memory
memory = ConversationBufferMemory(
    return_messages=True
)

# Use the same Ollama model
memory_llm = OllamaLLM(model="qwen2.5:3b")

conversation_prompt = PromptTemplate.from_template(
    """You are a helpful assistant.

Conversation history:
{history}

Human: {input}
Assistant:"""
)

memory_chain = conversation_prompt | memory_llm | StrOutputParser()

# Five conversation turns
turns = [
    "My name is Ughandhar.",
    "I am studying Robotics Engineering.",
    "What is my name?",
    "What am I studying?",
    "Can you remember both details about me?"
]

for i, user_input in enumerate(turns, 1):
    result = memory_chain.invoke({
        "history": memory.load_memory_variables({})["history"],
        "input": user_input
    })

    memory.save_context(
        {"input": user_input},
        {"output": result}
    )

    print(f"\n--- Turn {i} ---")
    print(f"Human: {user_input}")
    print(f"Assistant: {result}")

print("\n--- Conversation History ---")
for message in memory.load_memory_variables({})["history"]:
    print(f"{message.type}: {message.content}")
```
