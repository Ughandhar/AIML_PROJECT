```python
import pandas as pd

# Load the Indian cities dataset
df = pd.read_csv("../data/indian_cities.csv")

print("Dataset loaded successfully!")
print(df)
```

```python
# Inspect the dataset
print("Shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nFirst 10 rows:")
print(df.head(10))
```

```python
# Filter cities with literacy rate above 90%
high_literacy = df[df["literacy_rate"] > 90]

print(high_literacy)
```

```python
# Group cities by state and calculate average population
state_population = df.groupby("state")["population_millions"].mean()

print(state_population)
```

```python
# Create a second DataFrame for merge operation
city_category = pd.DataFrame({
    "city": ["Bengaluru", "Mumbai", "Delhi", "Chennai", "Hyderabad"],
    "category": ["IT Hub", "Financial Hub", "Capital", "Industrial Hub", "IT Hub"]
})

# Merge the city dataset with city categories
merged_df = pd.merge(df, city_category, on="city", how="left")

print(merged_df)
```

```python
# Create a pivot table showing average literacy rate by state
pivot = pd.pivot_table(
    df,
    values="literacy_rate",
    index="state",
    aggfunc="mean"
)

print(pivot)
```

```python
from pathlib import Path

# Create output directory
output_dir = Path("../data/output")
output_dir.mkdir(parents=True, exist_ok=True)

# Export DataFrame to CSV
csv_path = output_dir / "indian_cities_cleaned.csv"
df.to_csv(csv_path, index=False)

# Export DataFrame to Parquet
parquet_path = output_dir / "indian_cities_cleaned.parquet"
df.to_parquet(parquet_path, index=False)

# Compare file sizes
csv_size = csv_path.stat().st_size
parquet_size = parquet_path.stat().st_size

print(f"CSV file size: {csv_size} bytes")
print(f"Parquet file size: {parquet_size} bytes")
```
