```python
df = pd.read_csv("../data/indian_cities.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nSummary Statistics:")
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Top 5 observations:")

print("1. Highest population:", df.loc[df["population_millions"].idxmax(), "city"])
print("2. Lowest population:", df.loc[df["population_millions"].idxmin(), "city"])
print("3. Highest literacy rate:", df.loc[df["literacy_rate"].idxmax(), "city"])
print("4. Lowest literacy rate:", df.loc[df["literacy_rate"].idxmin(), "city"])
print("5. Highest temperature:", df.loc[df["average_temperature"].idxmax(), "city"])
```

```python
df.hist(figsize=(10, 6))
plt.tight_layout()
plt.show()
```

```python
correlation = df.select_dtypes(include="number").corr()

print(correlation)

sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
city_counts = df["city"].value_counts()

print("City Counts:")
print(city_counts)
```

```python
## EDA Observations

1. Delhi has the highest population at 33.8 million.
2. Kochi has the lowest population at 2.1 million.
3. Kochi has the highest literacy rate at 96.2%.
4. Jaipur has the lowest literacy rate at 76.4%.
5. Chennai has the highest average temperature at 29.0°C.

### Correlation Findings

- Population and literacy rate show almost no correlation (-0.039).
- Population and average temperature show a weak negative correlation (-0.26).
- Literacy rate and average temperature show a weak positive correlation (0.33).

### Data Quality

- No missing values were found.
- No duplicate rows were found.
- The dataset is ready for further analysis.
```
