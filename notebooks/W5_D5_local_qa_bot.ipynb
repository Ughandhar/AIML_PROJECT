# W5D5 — Local Q&A Bot with Ollama + ChromaDB

This notebook builds a local Q&A bot using Ollama and ChromaDB.

## System Prompt

The assistant should answer clearly and concisely using the provided context. It should avoid inventing information and should say when the answer is not available in the provided context.

```python
system_prompt = """
You are a helpful robotics and AI assistant.
Answer questions clearly and concisely.
Use the provided context when available.
Do not invent information.
If the answer is not available in the provided context, clearly say that you do not have enough information.
"""

print(system_prompt)
```

```python
def ask_ollama(question, model="llama3.2:3b"):
    response = ollama.chat(
        model=model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
        ]
    )
    
    return response["message"]["content"]
```

```python
question = "What is LiDAR used for in robotics?"

answer = ask_ollama(question)

print("Question:", question)
print("\nAnswer:", answer)
```

```python
test_prompts = [
    "What is LiDAR used for in robotics?",
    "What is the purpose of sensor fusion?",
    "Explain SLAM in simple terms.",
    "How does a robot avoid obstacles?",
    "What is the role of path planning in autonomous robots?"
]

for i, prompt in enumerate(test_prompts, 1):
    print(f"\n--- Prompt {i} ---")
    print("Question:", prompt)
    print("Answer:", ask_ollama(prompt))
```

```python
comparison_questions = [
    "What is LiDAR used for in robotics?",
    "What is sensor fusion and why is it useful?",
    "Explain SLAM in simple terms."
]

for i, question in enumerate(comparison_questions, 1):
    print(f"\n{'='*70}")
    print(f"QUESTION {i}: {question}")
    print(f"{'='*70}")

    llama_answer = ask_ollama(question, model="llama3.2:3b")
    qwen_answer = ask_ollama(question, model="qwen2.5:3b")

    print("\n--- llama3.2:3b ---")
    print(llama_answer)

    print("\n--- qwen2.5:3b ---")
    print(qwen_answer)
```

## Model Comparison — llama3.2:3b vs qwen2.5:3b

### Question 1 — LiDAR in Robotics
- **llama3.2:3b:** Gave a more detailed explanation with numbered applications such as autonomous vehicles, navigation, object recognition, and SLAM.
- **qwen2.5:3b:** Gave a shorter, focused explanation covering navigation, mapping, obstacle detection, and localization.

### Question 2 — Sensor Fusion
- **llama3.2:3b:** Provided a detailed explanation of combining multiple sensor inputs and its benefits.
- **qwen2.5:3b:** Provided a concise explanation focused on combining sensor information for better perception and reliability.

### Question 3 — SLAM
- **llama3.2:3b:** Explained SLAM with additional detail about mapping and localization.
- **qwen2.5:3b:** Gave a concise definition: building a map while simultaneously tracking the robot's location.

### Overall Observation
Both models produced relevant answers to the same robotics questions. The main difference observed was response style: llama3.2:3b generally produced more detailed explanations, while qwen2.5:3b tended to provide shorter and more concise responses.
