```python
import ollama

question = "Explain how a PID controller works in robotics."

system_prompts = {
    "Basic": "You are a helpful assistant.",
    
    "Role-based": (
        "You are a robotics engineering professor. "
        "Explain concepts accurately to an undergraduate engineering student."
    ),
    
    "Structured": (
        "You are a robotics expert. "
        "Explain the concept clearly using: Definition, Working, "
        "Example, and Key Points."
    )
}

for name, system_prompt in system_prompts.items():
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
        ]
    )
    
    print(f"\n===== {name} System Prompt =====")
    print(response["message"]["content"])
```

```python
import ollama

question = "Explain how a PID controller works in robotics."

system_prompts = {
    "Basic": "You are a helpful assistant. Answer in 40 words or less.",
    
    "Role-based": (
        "You are a robotics engineering professor. "
        "Explain accurately to an undergraduate student in 40 words or less."
    ),
    
    "Structured": (
        "You are a robotics expert. "
        "Answer using Definition, Working, and Example. "
        "Keep the complete answer under 40 words."
    )
}

for name, system_prompt in system_prompts.items():
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
        ]
    )

    answer = response["message"]["content"].replace("\n", " ")
    
    print(f"\n===== {name} System Prompt =====")
    print(answer)
```

```python
import ollama

question = "Explain how an autonomous robot avoids obstacles."

zero_shot = ollama.chat(
    model="llama3.2:3b",
    messages=[
        {
            "role": "user",
            "content": question
        }
    ]
)

few_shot = ollama.chat(
    model="llama3.2:3b",
    messages=[
        {
            "role": "user",
            "content": """
You are a robotics tutor.

Example:
Question: What is LiDAR?
Answer: LiDAR measures distances using laser pulses and helps robots understand their surroundings.

Example:
Question: What is ultrasonic sensing?
Answer: Ultrasonic sensing uses sound waves to measure distance and detect nearby objects.

Now answer:
Question: Explain how an autonomous robot avoids obstacles.
"""
        }
    ]
)

print("===== ZERO-SHOT =====")
print(zero_shot["message"]["content"])

print("\n===== FEW-SHOT =====")
print(few_shot["message"]["content"])
```

### Zero-Shot vs Few-Shot Prompting

Zero-shot prompting asks the model to answer a question without providing examples. In this experiment, the zero-shot response was detailed and included multiple sensors, algorithms, and obstacle-avoidance techniques.

Few-shot prompting provides examples before the actual question. The few-shot response followed the demonstrated structure and focused on common sensor-based obstacle avoidance methods.

**Observation:** Few-shot prompting can guide the model toward a more consistent response style, while zero-shot prompting allows the model to generate an answer without example-based guidance.

```python
import ollama

question = "Explain LiDAR in robotics."

formats = {
    "Short": "Answer in exactly 2 sentences.",
    "Bullet": "Answer using exactly 3 bullet points.",
    "Technical": "Answer in 3 sentences and include one technical term."
}

for name, instruction in formats.items():
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "system",
                "content": f"You are a robotics tutor. {instruction}"
            },
            {
                "role": "user",
                "content": question
            }
        ]
    )

    print(f"\n===== {name.upper()} FORMAT =====")
    print(response["message"]["content"])
```

### Prompt Constraint Experiment

The same LiDAR question was tested with three different output constraints: short, bullet-point, and technical formats.

**Observation:** The system prompt influenced the response style and structure. The bullet prompt produced a clear three-part answer, while the technical prompt produced a more detailed explanation using technical terminology. However, the model did not always follow strict constraints such as an exact sentence count.

**Key Learning:** Prompt constraints can guide LLM output format and level of detail, but they may not always be followed perfectly.

```python
import ollama

combined_prompt = """
You are a robotics engineering tutor. Give clear, technically accurate answers for undergraduate students.

Follow the style shown in these examples.

Example 1:
Question: What is LiDAR?
Answer: LiDAR uses laser pulses to measure distances and create a representation of the environment.

Example 2:
Question: What is an ultrasonic sensor?
Answer: An ultrasonic sensor uses high-frequency sound waves to measure distance and detect nearby objects.

Now answer:
Question: What is a camera sensor in robotics?

Keep the answer under 40 words.
"""

response = ollama.chat(
    model="llama3.2:3b",
    messages=[
        {
            "role": "user",
            "content": combined_prompt
        }
    ]
)

print(response["message"]["content"])
```

### Combined Prompting Experiment

A role-based instruction, few-shot examples, and an output-length constraint were combined in a single prompt.

**Observation:** The model followed the demonstrated answer style and produced a concise, technically relevant response. Combining multiple prompting techniques can provide better control over the content, style, and length of the generated response.

**Key Learning:** System instructions and few-shot examples can be combined to guide an LLM toward more consistent and controlled outputs.

```python
import ollama

questions = [
    "What is ROS in robotics?",
    "Explain how a PID controller works.",
    "What is LiDAR and why is it useful for robots?"
]

models = ["llama3.2:3b", "qwen2.5:3b"]

for question in questions:
    print("\n" + "=" * 70)
    print("QUESTION:", question)
    print("=" * 70)

    for model in models:
        response = ollama.chat(
            model=model,
            messages=[
                {
                    "role": "system",
                    "content": "You are a robotics engineering tutor. Give clear and technically accurate answers for undergraduate students."
                },
                {
                    "role": "user",
                    "content": question
                }
            ]
        )

        print(f"\n--- {model} ---")
        print(response["message"]["content"])
```

```python
answer = response["message"]["content"]
print(f"\n{model}:")
print(answer)
```
