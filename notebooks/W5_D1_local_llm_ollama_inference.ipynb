```python
import ollama

system_prompt = """
You are a helpful AI assistant specializing in robotics and artificial intelligence.
Give clear, concise, technically accurate answers suitable for an engineering student.
"""

prompts = [
    "What is ROS and why is it used in robotics?",
    "Explain the difference between supervised and unsupervised learning.",
    "What is a PID controller?",
    "How does LiDAR help an autonomous robot?",
    "What is the difference between AI and machine learning?"
]

for i, prompt in enumerate(prompts, start=1):
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": prompt}
        ]
    )

    print(f"\n--- Prompt {i} ---")
    print(f"Question: {prompt}")
    print(f"Response: {response['message']['content']}")
```

```python
import ollama

models = ["llama3.2:3b", "qwen2.5:3b"]

comparison_prompts = [
    "What is ROS and why is it used in robotics?",
    "Explain supervised vs unsupervised learning.",
    "What is a PID controller?",
    "How does LiDAR help an autonomous robot?",
    "What is the difference between AI and machine learning?"
]

for model in models:
    print(f"\n===== {model} =====")

    for i, prompt in enumerate(comparison_prompts, start=1):
        response = ollama.chat(
            model=model,
            messages=[
                {
                    "role": "system",
                    "content": "Answer accurately for an engineering student in 30 words or less."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        print(f"Prompt {i}: {response['message']['content'].replace(chr(10), ' ')}")
```

## Model Comparison — Llama 3.2 3B vs Qwen 2.5 3B

Both models produced relevant answers for all five robotics and AI questions.

**Llama 3.2:3b**
- Provided more detailed explanations.
- Gave more context for robotics concepts such as ROS and PID.
- Responses were generally more descriptive.

**Qwen 2.5:3b**
- Produced shorter and more direct answers.
- Focused on the core definition of each concept.
- Responses were concise and easy to scan.

**Overall observation**

For this small five-prompt comparison, Llama 3.2:3b tended to provide more detailed explanations, while Qwen 2.5:3b generally gave shorter responses. The comparison is qualitative and based only on these five test prompts, so it should not be treated as a comprehensive benchmark.
